```python
# 1. Створіть одновимірний масив (вектор) з простими числами, не перевищуючи 70, та виведіть його значення.
```

```python
import numpy as np
import random
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

print(np.array([x for x in range(2, 71) if is_prime(x)]))
```

```python
# 11. Створіть матрицю 2x3 та вектор 3x1 з випадковими дійсними числами від 0 до 1 і знайдіть суму елементів матриці, менших за відповідний елемент вектора
```

```python
# 14. Створіть матрицю 3x3 з випадковими дійсними числами від -5 до 5 та знайдіть добуток елементів, більших за середнє
```

```python
M = np.random.uniform(-5, 5, size=(3, 3))
mean_val = np.mean(M)
print("Матриця M:\n", M)
print("Середнє значення:", mean_val)
product = np.prod(M[M > mean_val])
print("Добуток елементів більших за середнє:", product)
```

```python
# 15. Створіть матрицю 3x3 з випадковими дійсними числами від -0.5 до 0.5 та знайдіть вектор-стовпчик з найбільшими за модулем цілими значеннями в кожному рядку
```

```python
M = np.random.uniform(-0.5, 0.5, size=(3, 3))
max_abs_int = np.array([int(np.round(np.max(np.abs(M[i])))) for i in range(M.shape[0])]).reshape(-1, 1)
print("Матриця M:\n", M)
print("Вектор-стовпчик найбільших за модулем цілих чисел по рядках:\n", max_abs_int)
```
